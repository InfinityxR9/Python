```python
import pandas as pd
import numpy as np
```

```python
movies = pd.read_csv('data/imdb-top-1000.csv')
movies.head(1)
```

```python
# Creates groups by categorial data (Genre)

genres = movies.groupby('Genre')
genres
```

```python
# Apply built-in aggregation functions

genres.mean(numeric_only=True)
# set numeric_only = True explicitly to calculate mean on numeric values only 
```

```python
# find the genre with highest avg IMDB rating

movies.groupby("Genre")['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

```python
# find director with most popularity (based on number of votes)

movies.groupby("Director")['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# find the highest rated movie of each genre (on the basis of imdb rating)

movies.groupby("Genre").sort_values('IMDB_rating').head(1)
```

```python
# find number of movies done by each actor 

# movies['Star1'].value_counts()

movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)
```
