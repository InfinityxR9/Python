```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = [0, 2, 1, 2, 0, 5, 5, 0, 1, 3, 2, 0, 1, 2, 0]
freq = [data.count(i) for i in data]
bin_list = [0,1,2,3,4, 5]
plt.hist(data, bins=bin_list, edgecolor='blue', color='skyblue', label='Histogram 1')
plt.legend(loc='best')  # you can also use frameon=False
plt.xlabel('data')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.grid(axis='both')
# print(freq)

plt.show()
```

```python
subjects = ['Chut', 'Gand', 'Lund', 'Sex', 'Bakchodi']
marks = [69, 20, 100, 60, 45]

plt.subplot(1,2,1)
plt.pie(marks, labels=subjects, autopct='%1.2f%%')
plt.title('Pie Chart')

plt.subplot(1,2,2)
plt.bar(subjects, marks)
plt.yticks(marks)
plt.grid()

plt.tight_layout()

plt.show()
```

```python
data = [
    [260, 320, 400, 350],
    [150, 240, 549, 980],
    [560, 350, 450, 200],
    [600, 450, 340, 650]
]

labels = ["Apple", "Banana", "Cherry", "DragonFruit"]
plt.boxplot(
    data, 
    tick_labels=labels,
    capprops=dict(color='purple', linewidth=1.5),
    boxprops=dict(color='black'),
    whiskerprops=dict(color='orange', linewidth=1),
    medianprops=dict(color='red', linewidth=1.5)
)

plt.title("Fruits bought by 4 customers")
plt.ylabel("Weight (in g)")
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

```

```python
data = [3, 4, 4, 4, 6, 7, 8, 8, 9, 10]

unique = sorted(set(data))
counts = [data.count(i) for i in unique]
plt.subplot(1, 2, 1)

plt.scatter(unique, counts, color='skyblue',
            marker='o', edgecolors='black', s=80)
plt.title("Scatter Plot freq of values")
plt.xlabel("Data Values")
plt.ylabel("Frequency")

sizes = [count*200 for count in counts]
colors = counts
plt.subplot(1, 2, 2)
plt.scatter(unique, counts, s=sizes, c=colors,
            cmap='plasma', alpha=0.6, edgecolors='black')
# can also choose cmap as viridis, magma, plasma, inferno, cividis
plt.title("Bubble Plot - Frequency of values")
plt.xlabel("Data Values")
plt.ylabel("Frequency")


plt.colorbar(label="Frequency")
plt.grid()
plt.show()
```

```python
domain = np.linspace(0, 2*np.pi, 100)
y = np.sin(domain)
# plt.grid()
plt.axvline(0, color='black', alpha=0.6, linestyle="--")
plt.axhline(0, color='black', alpha=0.6, linestyle="--")

xtick = np.linspace(0, 2*np.pi, 5)
plt.xticks(xtick)
plt.grid()

plt.plot(domain, y)
```
