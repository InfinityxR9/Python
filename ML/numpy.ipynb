## NumPy Arrays Vs Python List

```python
# speed

import time

# list 
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]

c = []

start = time.time()

for i in range(len(a)):
    c.append(a[i] + b[i])

print(time.time() - start)  # time taken for operation
```

```python
# NumPy arrays are build on C arrays, not dynamic arrays like python list

a = np.arange(10000000)
b = np.arange(10000000, 20000000)

start = time.time()
c = a + b
print(time.time() - start)
```

```python
1.0445449352264404/0.019994258880615234

# almost 50x faster (But no change in time complexity)
```

```python
# find numbers not divisible by 7

print(a[a % 7 != 0])

# or

# print(a[not (a % 7 == 0)])  # Same problem
print(a[~ (a % 7 == 0)])
```

## Broadcasting

The term broadcasting describes how numPy treats arrays of different shapes during arithmetic operations  
The smaller array is "broadcast" across the larger array so that they have compatible shapes

```python
# same shape

a = (np.random.random((2,3)))*50
b = (np.random.random((2,3)))*50

print(a, b, sep='\n')
print(a+b)
```

```python
# different shape

a = np.arange(6).reshape((2, 3))
b = np.arange(3).reshape(1, 3)

print(a)
print(b)
print("----")
print(a+b)
```

### Broadcasting Rules
**1. Make the two arrays have the same number of dimensions.**
- If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimension.<br>

**2. Make each dimension of the two arrays the same size.**
- If the sizes of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.
- If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.

<img src = "https://jakevdp.github.io/PythonDataScienceHandbook/figures/02.05-broadcasting.png">

```python
# More examples

a = np.arange(12).reshape(4, 3)
b = np.arange(3)

print(a)
print(b)

print(a+b)
```

```python
a = np.arange(12).reshape(3,4)
b = np.arange(3)

# ** Only dim(1) can be extended/streched for broadcasting

print(a)
print(b)

print(a+b)
```

## Working with Mathematical Formula

```python
# Sigmoid function

# (In general) a function which maps every number in its domain with a number within (0, 1)
# (Here) takes an array and maps every element to some number betwee 0 and 1

def sigmoid (arr: np.ndarray):
    return 1/(1+np.exp(-(arr)))

sigmoid(np.array([1,2,3]))
```

```python
# Mean squared Error 

# used in Linear Regression loss function 
# defined as the mean of square of errors (loss from actual value)

def MSE (actual:np.ndarray, predicted:np.ndarray):
    return np.mean(np.pow(actual - predicted, 2))

a = np.random.randint(1, 50, 25)
b = np.random.randint(1, 50, 25)
print(a)
print(b)
MSE(a, b)
```

```python
# binary cross entropy

# used for Logistic regression Loss function

def BCE(actual: np.ndarray, predicted: np.ndarray):
    return -np.mean(predicted*np.log(predicted) + (1-actual)*np.log(1-predicted))

a = np.random.random(10)
b = np.random.random(10)
print(a)
print(b)

BCE(a, b)
```
