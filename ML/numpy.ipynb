```python
import numpy as np
```

## Creating numpy arrays

```python
a = np.array([1,2,3])
print(type(a), a)   # Vector
```

```python
# 2D and 3D
b = np.array([[1,2,3], [4,5,6]])    # Matrix
print(b)

c = np.array([[[1,2], [4,5]], [[3,4], [6,7]]])
print(c)    # Tensor

d = np.array([1,2,3], dtype=float)
e = np.array([1,2,3], dtype=complex)
f = np.array([1,2,3], dtype=bool)
print(d, e, f, sep='\n')
```

```python
# np.arange()

# Just like the classic range(), just float values would also be allowed

print(np.arange(3.4, 5, 0.1))
```

```python
# np.reshape()

# very much used with arange() to convert the 1D array into some other dimension

print(np.arange(1, 10).reshape(3, 3))
print(np.arange(-10, 14).reshape(2, 2, 2, 3))
# ValueError if it is not shape-able
```

```python
# np.zeros() and np.ones()

# np.zeros() ==> arg: shape, initialise an array with given shape with zereos
# np.zeros() ==> arg: shape, initialise an array with given shape with ones

# np.random.random() ==> arg: shape, initialise an array with given shape with random numbers

# default dtype is float, use can change it to dtype = int

print(np.zeros((2,3)))
print(np.ones((2,3), dtype=int))
print(np.random.random((2,3)))
```

```python
# np.linspace() ==> Linear Space
# np.linspace() ==> arg: start, end, number of values... creates the points at equal distance

# 5 numbers between -10 and 10 such that they are equally spaced
print(np.linspace(-10, 10, 5))
```

```python
# np.identity()   ==> makes an identity matrix of given order 

print(np.identity(3))
print(np.identity(4))
```

## Numpy Array Attributes

```python
a1 = np.arange(10,dtype=np.int32)
a2 = np.arange(12,dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)    # by default int64 (takes 8 bytes in memory, same as float)
```

```python
# ndim ==> returns number of dimensions

print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
# shape

# can be intepreted as for n dim array: (...): first element how many (n-1) dim arrays, and then in each one of them, how many (n-2) dim arrays by following element and so on...

print(a1.shape) # (10,) as there're 10 0D arrays (single element) with no -1D arrays in it
print(a2.shape) # (3, 4) as there're 3 1D arrays with 4 0D arrays (single element) in it
print(a3.shape) # (2, 2, 2): there're 2 2D arrays, in each, 2 1D arrays and in each 2 elements
```

```python
# size

# gives number of elements

print(a1.size)
print(a2.size)
print(a3.size)
```

```python
# itemsize

# gives size of each item in memory (in bytes)

print(a1.itemsize)  # each item 32bit int
print(a2.itemsize)  # each item 64bit float
print(a3.itemsize)  # each item 64bit int
```

```python
# dtype

# gives the datatype

print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

## Changing DataType

```python
# astype

print(a2.astype(np.int32))
print(a2)   # doesn't change the original array!
```

## Array Operations

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)
```

```python
# Scalar Operations

# arithmetic
print(a1*2)
pow(a1, 2)
print(a2>15)
print(a3==15)
```

```python
# Vector Operations

a1 ** a2    # correspondingly perform operations (ValueError if shapes are different)
```

## Array Functions

```python
a1 = np.random.random((3,3))*100
a1
```

```python
# max/min/sum/prod

print(np.max(a1))
print(np.min(a1))
print(np.prod(a1))
print(np.sum(a1))

# 0 --> col; 1--> row 
print(np.max(a1, axis=0))   # Gives max col wise
# Applicable to all those functions
```

```python
# mean/median/std/var 

np.mean(a1, axis=0)     # You got it!
```

```python
# trigo functions

np.sin(a1)
```

```python
# dot product ==> same as matrix multiplication

a2 = np.arange(12).reshape(3, 4)
a3 = np.arange(12, 24).reshape(4, 3)

np.dot(a2, a3)
```

```python
# log and exp 

print(np.log(a1))
np.exp(a2)
```

```python
# round/floor/ceil

# round => rounds off to nearest integer
# floor => GIF (round off to prev. integer (greatest lower integer bound))
# ceil => round off to next integer (lowest greater integer bound)

np.int32(np.ceil(a1))
```

## Indexing and Slicing

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)

a3
```

```python
a1
```

```python
# indexing: exactly same as python list indexing

print(a1[-1])
```

```python
print(a2[1][2])

# or simply

print(a2[1,2])
print(a2[1,0])
```

```python
a3
```

```python
print(a3[1,0,1])
print(a3[0, 0, 0])
print(a3[0, 1, 0])
```

```python
# Slicing : same as python list slicing

print(a1[2:5:2])
a2
```

```python
print(a2[0, :])     # first row, all columns
print(a2[:, 2])     # all rows, third columns
print(a2[1:, 1:3])
print(a2[::2, ::3]) # gets the corner items
print(a2[::2, 1::2])
print(a2[1, ::3])
print(a2[:2, 1::])
```

```python
a3 = np.arange(27).reshape(3,3,3)
a3
```

```python
print(a3[1])
print(a3[::2])
print(a3[0, 1])
print(a3[1, :, 1])
print(a3[::2, 0, ::2])
```

## Stacking

Concate arrays over each other  
The shape of arrays should be same

```python
# horizontal stacking
a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)
a5
```

```python
np.hstack((a4, a5))
```

```python
# Vertical stacking

np.vstack((a4, a5))
```

## Splitting

Opposite of stacking  
hsplit --> splitting horizontally (vertical cut)  
vsplit --> splitting vertically (horizontal cut)

```python
a4
```

```python
np.hsplit(a4, 4)    # has to be equal division (Otherwise ValueError)
```

```python
a5
```

```python
np.vsplit(a5, 3)
```
