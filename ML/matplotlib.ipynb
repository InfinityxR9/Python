## 2D Line plot

image.png

- Used for Bivariate Analysis
- Used with categorial & numerical datas or both numerical datas
- The most use case is Time Series data

```python
# plotting a simple function 

price = [48000, 46000, 50000, 55000, 47000, 52000]
year = [2015, 2016, 2017, 2018, 2019, 2020]

plt.plot(year, price)
```

```python
# from pandas dataframe

batsman = pd.read_csv('data/sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
# plotting multiple plots with colors, linestyle (dotted, dashed, dash-dott), linewidth
# use markers for (x, y) points (d for diamond, +, .), markersize
# label to give individual plot a label

plt.plot(batsman['index'], batsman['V Kohli'],
         color='black', linestyle='--', linewidth=0.75, marker='.', markersize=10, label="V Kohli")
plt.plot(batsman['index'], batsman['RG Sharma'],
         color="#72f8cb", linewidth=2, marker='^', markersize=10, label="R Sharma")

# You can also give hex codes for color

# labels title

plt.title('V Kohli VS R Sharma')

# axes label

plt.xlabel('Season')
plt.ylabel('Runs scored')

# use plt.legend() to show the labels
plt.legend()
# use loc parameter to give the location of box (default is best) (right upper,..etc...)
# you can also use frameon=False to hide the box frame on legend

# To show the grids
plt.grid()

# to specify grid lines at x axis and label at those ticks xticks(ticks location, label)
plt.xticks(batsman['index'])

# to specify grid lines at y axis 
# plt.yticks([*batsman['RG Sharma'], *batsman['V Kohli']])

plt.show()
# To show all the plots
```

```python
# Limiting Axes

price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
# This would appear flat because the last data point is too large in compare to other points

plt.ylim(0, 75000)

plt.grid()
```

## Scatter Plots

image.png

- Bivariate Analysis
- numerical vs numerical
- Use case - Finding correlation

```python
# plt.scatter simple function
x = np.linspace(-10,10,50)

y = 10*x + 3 + np.random.randint(0,300,50)
y
```

```python
plt.scatter(x,y)
```

```python
# On pandas dataset

df = pd.read_csv('data/batter.csv')
df = df.head(50)
df
```

```python
plt.scatter(df['avg'], df['strike_rate'], marker='*', c=df['avg'], cmap='magma')
plt.colorbar(label="Average Score")
plt.title('Average and SR Analysis of Top 50 batsman')
plt.xlabel("Average")
plt.ylabel("Strike Rate")
```

```python
# size

tips = sns.load_dataset('tips')
tips
```

```python
# size

plt.scatter(tips['total_bill'], tips['tip'], s=tips['size']*10)
```

```python
# scatterplot using plt.plot

plt.plot(tips['total_bill'], tips['tip'], 'o')
```

```python
# scatter plot using plt.plot() is faster than plt.scatter() function but with less customisation 
```

## Bar Chart

image.png

- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [10, 20, 40, 10, 30]
colors = ['red', 'blue', 'green', 'yellow', 'pink']

plt.bar(colors, children, color=colors)
```

```python
# simple horizontal bar chart
children = [10, 20, 40, 10, 30]
colors = ['red', 'blue', 'green', 'yellow', 'pink']

plt.barh(colors, children, color=colors)
```

```python
# Multiple bar graphs
df = pd.read_csv('data/batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0])-0.3, df['2015'], width=0.3, label='2015')
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.3, label='2016')
plt.bar(np.arange(df.shape[0])+0.3, df['2017'], width=0.3, label='2017')
plt.legend()

# Sometimes the category labels might be too long and would overlap
# to solve this use, rotation parameter in xticks
plt.xticks(np.arange(df.shape[0]), df['batsman'], rotation=90)

plt.show()
```

```python
# Stacked Bar chart
# use bottom parameter to specify the starting coordinate of the bars 

plt.bar(df['batsman'], df['2015'], label="2015", width=0.5)
plt.bar(df['batsman'], df['2016'], bottom=df['2015'], label="2016", width=0.5)
plt.bar(df['batsman'], df['2017'], bottom=df['2016']+df['2015'], label="2017", width=0.5)

plt.legend()
```

## Histogram

image.png

- Univariate Analysis
- Numerical col
- Use case - Frequency Count

```python
df = pd.read_csv('data/vk.csv')
df
```

```python
plt.hist(df['batsman_runs'], bins=np.arange(0,100,5), rwidth=0.75)
plt.xticks(np.arange(0,100,5))

plt.xlabel('Batsman Runs')
plt.ylabel('Number of matches')

plt.grid()

plt.show()
```

```python
# Logarithmic Scale

arr = np.load('data/big-array.npy')
plt.hist(arr, bins=np.arange(10,80,10), log=True, rwidth=0.75)

plt.xlabel('Values')
plt.ylabel("Frequency")
```

## Pie Chart

image.png

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
# simple data
data = [23, 45, 100, 20, 49]
subjects = ['eng', 'science', 'maths', 'sst', 'hindi']
plt.pie(data, labels=subjects, autopct="%1.2f%%")
plt.title("Marks distribution")

plt.show()
```

```python
df = pd.read_csv('data/gayle-175.csv')
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%1.2f%%')
plt.title("Batsman Contribution")
```

```python
# Explode and shadow

plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%',
        explode=[0.3, 0, 0, 0, 0, 0.1], shadow=True)
plt.show()
```
