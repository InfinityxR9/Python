```python
import matplotlib.pyplot as plt
import numpy as np

data = [
    [260, 320, 400, 350],
    [150, 240, 549, 980],
    [560, 350, 450, 200],
    [600, 450, 340, 650]
]

labels = ["Apple", "Banana", "Cherry", "DragonFruit"]
plt.boxplot(
    data, 
    tick_labels=labels,
    capprops=dict(color='purple', linewidth=1.5),
    boxprops=dict(color='black'),
    whiskerprops=dict(color='orange', linewidth=1),
    medianprops=dict(color='red', linewidth=1.5)
)

plt.title("Fruits bought by 4 customers")
plt.ylabel("Weight (in g)")
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt
data = [3, 4, 4, 4, 6, 7, 8, 8, 9, 10]

unique = sorted(set(data))
counts = [data.count(i) for i in unique]
plt.subplot(1, 2, 1)

plt.scatter(unique, counts, color='skyblue',
            marker='o', edgecolors='black', s=80)
plt.title("Scatter Plot freq of values")
plt.xlabel("Data Values")
plt.ylabel("Frequency")

sizes = [count*200 for count in counts]
colors = counts
plt.subplot(1, 2, 2)
plt.scatter(unique, counts, s=sizes, c=colors,
            cmap='viridis', alpha=0.6, edgecolors='black')
plt.title("Bubble Plot - Frequency of values")
plt.xlabel("Data Values")
plt.ylabel("Frequency")


plt.colorbar(label="Frequency")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
male = [40, 80, 65, 35, 85, 30]
female = [50, 75, 50, 57, 35, 63]
label = ["Whatsapp", "insta", "yt", "kdrama", "chrome", "snapchat"]


plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.pie(male, labels=label, autopct='%1.2f%%', colors=[
        'green', 'blue', 'skyblue', 'red', 'magenta', 'purple'])
plt.title("Males Data")

plt.subplot(1, 2, 2)
plt.pie(female, labels=label, autopct='%1.2f%%', colors=[
        'green', 'blue', 'skyblue', 'red', 'magenta', 'purple'])
plt.title("Female Data")

plt.show()
```

```python
width = 0.3
x = [i + width for i in range(len(label))]

plt.bar(label, male, label="Male", width=0.3)
plt.bar(x, female, label="Female", width=0.3)
plt.legend()
plt.show()
```

```python
freq = np.array([1,2,3,4,5,6,5,4,3,2,1])
prob = np.array([i/np.sum(freq) for i in freq])

random_variable = np.arange(2,13)

plt.bar(random_variable, prob, width=1, edgecolor='blue', color='skyblue')
plt.xticks(random_variable)
plt.yticks(prob)
plt.plot(random_variable, prob)
plt.grid()
plt.show()
```
