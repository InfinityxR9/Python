## Vectorised String Operations

```python
# What are vectorised string operations?

a = np.array([1,2,3,4])
a*4
# This is a vectorised operation, we apply operation on each element of vector (numpy ndarray in this case)
```

```python
# Problems in vanilla python

s = ['cat', 'mat', None, 'rat']

[i.startswith('c') for i in s]
```

```python
# How pandas solve this issue...

s = pd.Series(['cat', 'mat', None, 'rat'])

# string accessor = str 
s.str.startswith('c')

# handles None values, fast and optimised for larger datasets
```

```python
df = pd.read_csv('data/titanic.csv')
df
```

```python
# Common Funtions
# lower/upper/capitalise/title 

df['Name'].str.title()
```

```python
# len 

df['Name'][df['Name'].str.len() == df['Name'].str.len().max()].values[0]
```

```python
# strip

df['Name'].str.strip()
```

```python
# split

df['Last Name'] = df['Name'].str.split(',').str.get(0)
# get method to get index position value 

df.head()
```

```python
df[['title', 'first name']] = df['Name'].str.split(',').str.get(1).str.split(n=1, expand=True)
# perform split only once
# expand=True makes it into a dataframe
df.head()
```

```python
df['title'].value_counts()
```

```python
# replace

df['title'] = df['title'].str.replace('Ms.', 'Miss.')
df['title'] = df['title'].str.replace('Mlle.', 'Miss.')

df['title'].value_counts()
```

```python
# filtering

# startswith/endswith
df[df['first name'].str.startswith('A')]

# isalpha/isalphanum/isdigit
```

```python
# Applying regex

# search john -> both cases 

df[df['first name'].str.contains('john', case=False)].head()
```

```python
# find passengers whose surname ends/starts with a vowel (Using RegEX)

df[df['Last Name'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]
```

```python
# slicing on each item

df['Name'].str[::-1]
```
