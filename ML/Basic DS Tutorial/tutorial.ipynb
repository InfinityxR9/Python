```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

img = mpimg.imread("image.png")
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
print(img)
```

```python
print(img.shape)
img2 = img[:, :, :1]
print(img2.shape)
plt.imshow(img2, cmap='gray')
plt.axis("off")
plt.show()
```

```python
img3 = img2.reshape(648, -1)
print(img3.shape)
plt.imshow(img2, cmap='gray')
plt.axis("off")
plt.show()
```

```python
img4 = img3[: : -1]
plt.imshow(img4, cmap='gray')
plt.axis("off")
plt.show()
```

```python
img5 = img3[: : -1, : : -1]
plt.imshow(img5, cmap='gray')
plt.axis("off")
plt.show()
```

```python
img6 = img3.T
plt.imshow(img6, cmap='gray')
plt.axis("off")
plt.show()
```

random number generation

```python
from IPython.display import clear_output

for i in range(100):
    arr = np.random.randint(0, 20, 40)
    
    clear_output(wait=True)
    plt.scatter(range(len(arr)), arr)
    plt.xlabel("Index")
    plt.ylabel("Value")
    plt.show()
    
```
