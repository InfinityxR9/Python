```python
import numpy as np
import matplotlib.pyplot as plt
import math

x = np.arange(-math.pi*1.5, math.pi*1.5, 0.01)
y = [math.sin(i) for i in x]
fig = plt.figure()
plt.plot(x, y, label='sine wave')
plt.axhline(y=0, color='blue')
plt.axhline(y=1, color='red', linestyle='--', linewidth=1.5)
plt.axhline(y=-1, color='red', linestyle='--', linewidth=1.5)
plt.axvline(x=0, color='blue')
plt.grid()

z = [math.cos(i) for i in x]
plt.plot(x, z, label='cosine wave')

plt.legend(loc='best')
plt.title("Wave functions")
plt.xlabel("Domain")
plt.ylabel("Function value")
plt.show()
plt.close(fig)
```
